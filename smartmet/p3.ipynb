# 步骤1：下载ERA5多年度夏季数据

下载 2010–2022 年夏季（6–8月）ERA5 单层数据，包含以下变量：
- **2m_temperature** — 地表2米气温
- **geopotential** — 位势（用于计算500 hPa位势高度）
- **sea_surface_temperature** — 海表温度
- **total_precipitation** — 总降水量
- **surface_pressure** — 地表气压
- **u10, v10** — 10米风场 U/V 分量

逐年下载，避免单次请求过大导致超时。

```python
import cdsapi
import os
from datetime import datetime

client = cdsapi.Client()

dataset = "reanalysis-era5-single-levels"

years = [str(y) for y in range(2010, 2023)]
months = ["06", "07", "08"]
days = [f"{d:02d}" for d in range(1, 32)]
times = ["00:00", "06:00", "12:00", "18:00"]

# 中国区域：[北纬, 西经, 南纬, 东经]
area = [50, 80, 20, 130]

variables = [
    "2m_temperature",
    "geopotential",
    "sea_surface_temperature",
    "total_precipitation",
    "surface_pressure",
    "10m_u_component_of_wind",
    "10m_v_component_of_wind",
]

output_dir = "era5_summer_data"
os.makedirs(output_dir, exist_ok=True)

for year in years:
    filename = f"{output_dir}/china_summer_{year}.nc"
    if os.path.exists(filename):
        print(f"{year} 已存在，跳过")
        continue

    print(f"\n=== 开始下载 {year} 年夏季数据 ===")
    request = {
        "product_type": "reanalysis",
        "variable": variables,
        "year": year,
        "month": months,
        "day": days,
        "time": times,
        "area": area,
        "format": "netcdf",
    }

    try:
        client.retrieve(dataset, request, filename)
        print(f"{year} 年下载完成 → {filename}")
    except Exception as e:
        print(f"{year} 年下载失败: {e}")

print("\n所有年份下载完毕！")
```

```python
# 用OpenCode生成这段，5分钟出第一个动画
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import numpy as np

# 读数据
ds = xr.open_dataset('temp.nc')

# 画逐日温度变化动画
for i, t in enumerate(ds.valid_time[:100]):  # 先做10帧试试
    fig = plt.figure(figsize=(10, 6))
    ax = plt.axes(projection=ccrs.PlateCarree())
    
    temp = ds['t2m'].sel(valid_time=t) - 273.15
    ax.contourf(temp.longitude, temp.latitude, temp, 
                levels=20, cmap='RdYlBu_r')
    ax.coastlines()
    ax.set_title(f"2m Temperature - {str(t.values)[:10]}")
    
    plt.savefig(f'frames/frame_{i:04d}.png', dpi=150)
    plt.close()

# 然后用ffmpeg合成：
# ffmpeg -framerate 5 -i frames/frame_%04d.png -c:v libx264 -pix_fmt yuv420p output.mp4
```
