```python
import string
db = ['Bob2049', 'Alice2026']
while True:
    name:str = input('Enter username:')
    if len(name) <6:
        print('must be at least six charactors.')
        continue
    if name[0] not in string.ascii_letters:
        print('must begin with alphabets.')
        continue
    if name.isalnum() and any(c.isdigit() for c in name) and any(c.isalpha() for c in name):
        print('ok')
        db.append(name)
        break
    else:
        print('must be alphabets and numbers')
print('users:', db)
```

```python
scores = ((88, 92, 55, 60, 71),(72, 66, 95, 81),(93, 89, 79, 86, 99, 85, 60))
def unpack(tur: tuple):
    ans = 0
    acco = 0
    for i in tur:
        if not isinstance(i, tuple):
            ans+=i
            acco+=1
        else:
            ans += unpack(i)[0]
            acco+=unpack(i)[1]
    return ans, acco
ans, acco = unpack(scores)
print(f'{ans/acco:.2f}')
```

```python
# 函数main定义
# 主体代码逻辑
def main():
    a, b = input().split()
    ans1 = alpha_str_gen(a, b)
    print(ans1)

    ans2 = alpha_str_gen(a, b, 2)
    print(ans2)
    

# 函数alpha_str_gen定义
# 待补足。。。
def alpha_str_gen(a, b, c=1):
    ans = ''
    for i in range(ord(a), ord(b)+1, c):
        ans+=chr(i)
    return ans
        


# 函数调用
main()
```

```python
# 递归函数comb定义
# 功能：计算从n个不同元素中取出m个的组合数C(n,m)
# 待补足...
def comb(n, m):
    if m == 0 or m == n:
        return 1
    elif m > n:
        return 0
    else:
        return comb(n-1, m-1) + comb(n-1, m)

# 测试代码
n, m = map(int,input().split())
print(comb(n, m)) # 函数调用
```
