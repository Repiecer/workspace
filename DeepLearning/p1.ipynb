```python
import torch
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
print(x)
```

```python
x = torch.Tensor([2])
w = torch.randn(1, requires_grad=True)
b = torch.randn(1, requires_grad=True)
y = torch.mul(w, x)
z = torch.add(y, b)
print(w.grad, b.grad)
```

```python
z.backward()
print(w.grad, b.grad)
```
